```python
indicators = {
    "GDP": "NY.GDP.MKTP.CD",
    "GDP_growth": "NY.GDP.MKTP.KD.ZG",
    "GDP_per_capita": "NY.GDP.PCAP.CD",
    "Inflation": "FP.CPI.TOTL.ZG",
    "Unemployment": "SL.UEM.TOTL.ZS",
    "Exports": "NE.EXP.GNFS.CD",
    "Imports": "NE.IMP.GNFS.CD",
    "Investment": "NE.GDI.TOTL.ZS"
}

print(indicators)
```

```python
data = {}

for name, indicator in indicators.items():
    result = wbdata.get_data(
        indicator,
        country="CIV"
    )
    data[name] = result

print(data.keys())
```

```python
indicator_codes = {
    code: name
    for name, code in indicators.items()
}

print(indicator_codes)
```

```python
data = wbdata.get_dataframe(
    indicator_codes,
    country="CIV",
    date=("2010", "2025")
)

print(data.head())
```

```python
print("Dataset shape:")
print(data.shape)

print("\nMissing values:")
print(data.isna().sum())

print("\nData types:")
print(data.dtypes)
```

```python
data.describe()
```

```python
print("Duplicate rows:", data.duplicated().sum())

print("\nIndex name:")
print(data.index.name)

print("\nIndex type:")
print(data.index.dtype)

print("\nYears:")
print(data.index.tolist())
```

```python
data.index = data.index.astype(int)
data = data.sort_index()

print(data.head())
```
